# Introduction

Notebook ini berisi script untuk membuat semua model untuk semua jenis product menggunakan auto arima

# Libraries

```python
import warnings
warnings.filterwarnings('ignore')

import joblib
import pandas as pd
from pmdarima import auto_arima
```

# Load Data

```python
file_path = '../data/pangoly.csv'
raw = pd.read_csv(file_path)

len(raw)
```

# Prepare Data

```python
data = raw.copy()
```

```python
data.dtypes
```

```python
data['date'] = pd.to_datetime(data['date'])
```

```python
dfs = {}

for cat in data['category'].unique():
    dfs[cat] = {}

    cat_df = data[data['category'] == cat]
    for prod in cat_df['product'].unique():
        df = cat_df[cat_df['product'] == prod].copy()
        df = df.set_index('date')
        df.index = df.index.normalize()
        df = df.sort_index()
        dfs[cat][prod] = df
```

# Models

```python
models = {}

for cat, prod_data in dfs.items():
    models[cat] = {}

    for prod, df in prod_data.items():
        model = auto_arima(df['avg'], seasonal=False, stepwise=True)
        models[cat][prod] = model
```

# Save Model

```python
# Result size = 325MB compressed 
# joblib.dump(models, '../models/arima/auto_arima.joblib', compress=3)
```
